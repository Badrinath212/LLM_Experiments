```python
import os
from dotenv import load_dotenv
from openai import OpenAI

load_dotenv(override=True)
```

```python
OPEN_ROUTER_API_KEY = os.getenv("OPEN_ROUTER_API_KEY")

if not OPEN_ROUTER_API_KEY:
    raise ValueError("OPEN_ROUTER_API_KEY environment variable is not set")
else: 
    print("OPEN_ROUTER_API_KEY environment variable is set")
```

```python
SYSTEM_PROMPT = "Your are an helpful assistant"
```

```python
USER_PROMPT = "Tell me about virat kohli"
```

```python
open_router = OpenAI(api_key=OPEN_ROUTER_API_KEY, base_url="https://openrouter.ai/api/v1")
```

```python
messages = [
    {"role": "system", "content": SYSTEM_PROMPT},
    {"role": "user", "content": USER_PROMPT}
]
```

```python
response = open_router.chat.completions.create(
    model="openrouter/free",
    messages=messages,
    temperature=0.7,
    max_tokens=1000,
)
```

```python
print(response.choices[0].message.content)
```

```python
from pydantic import BaseModel, Field

class Player(BaseModel):
    name: str = Field(description="The name of the player")
    age: int = Field(description="The age of the player")
    height: float = Field(description="The height of the player in meters")
    weight: float = Field(description="The weight of the player in kilograms")
    century: int = Field(description="The number of centuries scored by the player")
```

```python
from langchain_openai import ChatOpenAI

llm = ChatOpenAI(
    model_name="openrouter/free",
    api_key=OPEN_ROUTER_API_KEY,
    base_url="https://openrouter.ai/api/v1"
)
```

```python
structured_llm = llm.with_structured_output(Player)
```

```python
response = structured_llm.invoke("Tell me about virat kohli")
response
```
