```python
def get_customer_details(customer_id):
    for customer in customers_details:
        if customer['id'] == customer_id:
            return customer
    return None
```

```python
def search_orders(customer_id):
    orders = []
    for order in orders_details:
        if order['customer_id'] == customer_id:
            orders.append(order)
    return orders
```
